## Ficha das regras (Etapa 3)

Perfil dos balancetes de 2024 e 2025, conferido em 04/11/2026. Cada regra do `schemas.py` e a evidência que a justifica.

### Despesa (5.884 linhas, 41 colunas; 13 declaradas)

| Coluna | Regra | Evidência |
|---|---|---|
| `ANO` | int, `ge=2000` | só 2024 e 2025; sem teto para aceitar anos novos |
| `MES` | int, `eq=12` | só 12: os valores são o acumulado até dezembro |
| `COD_MUNICIPIO` | int, `eq=144` | só 144 (Pelotas) |
| `CD_ORGAO` | int, `eq=54400` | só 54400 (PM de Pelotas); trava pelo código, não pelo nome |
| `CD_FUNCAO` | str, `^\d{1,2}$` | 1 ou 2 dígitos ("4" a "99"), sem zero à esquerda |
| `CD_SUBFUNCAO` | str, `^\d{2,3}$` | 2 ou 3 dígitos ("62" a "999") |
| `CD_ELEMENTO` | str, `^\d\.\d\.\d{2}\.\d{2}$` | sempre no formato 3.1.90.04 |
| `DS_FUNCAO`, `DS_SUBFUNCAO` | str, sem vazio | nenhum vazio; a grafia muda entre os anos e é padronizada na transformação |
| `VL_DOTACAO_INICIAL` | float, `ge=0` | nenhum negativo |
| `VL_EMPENHADO`, `VL_LIQUIDADO`, `VL_PAGO` | float, `ge=0` | nenhum negativo |
| entre colunas | pago ≤ liquidado ≤ empenhado | nenhuma linha quebra a ordem |
| (regra descartada) | empenhado ≤ dotação inicial | reprovaria 1.736 linhas: a dotação muda no ano com créditos e remanejamentos |

### Receita (1.351 linhas, 24 colunas; 10 declaradas)

| Coluna | Regra | Evidência |
|---|---|---|
| `ANO`, `MES`, `COD_MUNICIPIO`, `CD_ORGAO` | iguais às da despesa | mesmos valores |
| `CD_CONTA_SG` | str, `^\d{15,16}$` | 15 dígitos; 16 só em 2024 (ramo das deduções do FUNDEB) |
| `DS_CONTA_SG` | str, sem vazio | nenhum vazio |
| `NIVEL` | str, `isin=["S", "A"]` | só S (agrupadora) e A (ponta) |
| `NR_NIVEL` | int, de 1 a 9 | de 1 a 9 nos dois anos; teto observado, não normativo |
| `VL_ORCADO`, `VL_ARRECADADO` | float, negativo permitido | 22 orçados e 14 arrecadados negativos: deduções do FUNDEB |
| (regra descartada) | `CD_CONTA_SG` único | o mesmo código aparece em várias linhas (por unidade e `CAR_PECULIAR`) |
| (fora do schema) | `NOME_UNIDADE` e afins | vazias em 458 linhas; a pipeline não usa |

### Decisões
- `strict = False`: os arquivos têm mais colunas do que as declaradas, e uma coluna nova da fonte não deve quebrar a pipeline.
- Validação com `Schema.validate(df)` depois da extração, e não com `@pa.check_output`: a `extract_all` devolve despesa ou receita, e o decorador só aceita um schema fixo.
- Município e órgão travados: o schema descreve o escopo de hoje; ampliar o escopo é mudar o schema de propósito.
